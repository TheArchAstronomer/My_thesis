```python
import matplotlib.pyplot as plt
import numpy as np
import astropy.units as u
import matplotlib as mpl
```

```python
m_lmc = 1.8e11 * u.Msun
r_sh = 20 * u.kpc

for i in range(1, 4):
    r = np.logspace(-2, 2.5, 100) * u.kpc * i

rho_h = (m_lmc / (2 * np.pi))*(r_sh / r) * (1 / (r + r_sh)**3)

m_lmc2 = 1.0e10 * u.Msun
rho_h2 = (m_lmc2 / (2 * np.pi))*(r_sh / r) * (1 / (r + r_sh)**3)

m_lmc3 = 2.0e11 * u.Msun
rho_h3 = (m_lmc3 / (2 * np.pi))*(r_sh / r) * (1 / (r + r_sh)**3)   

m_lmc4 = 5.0e7 * u.Msun
rho_h4 = (m_lmc4 / (2 * np.pi))*(r_sh / r) * (1 / (r + r_sh)**3)

m_lmc5 = 1.0e8 * u.Msun
rho_h5 = (m_lmc5 / (2 * np.pi))*(r_sh / r) * (1 / (r + r_sh)**3)

m_lmc6 = 5.0e8 * u.Msun
rho_h6 = (m_lmc6 / (2 * np.pi))*(r_sh / r) * (1 / (r + r_sh)**3)
```

```python
rs = 1.62 * u.kpc * (m_lmc / (1e8 * u.Msun))**(1/2)

rho_p = ((3 * m_lmc) / (4 * np.pi * rs**3)) * (1 + (r**2 / rs**2))**(-5/2)
```

```python
plt.rc('text', usetex=False)
plt.rc('font', family='serif')
fig = plt.figure(figsize=(8, 6))
#plt.plot(r, rho_p, label='Plummer (core) profile', linewidth=2)
plt.plot(r, rho_h4, label=r'Hernquist (cusp) profile, $5 \cdot 10^{7} M_\odot$', linewidth=2)
plt.plot(r, rho_h5, label=r'Hernquist (cusp) profile, $1 \cdot 10^{8} M_\odot$', linewidth=2)
plt.plot(r, rho_h6, label=r'Hernquist (cusp) profile, $5 \cdot 10^{8} M_\odot$', linewidth=2)
plt.plot(r, rho_h2, label=r'Hernquist (cusp) profile, $1 \cdot 10^{10} M_\odot$', linewidth=2)
plt.plot(r, rho_h, label=r'Hernquist (cusp) profile, $1.8 \cdot 10^{11} M_\odot$', linewidth=2)
plt.plot(r, rho_h3, label=r'Hernquist (cusp) profile, $2 \cdot 10^{11} M_\odot$', linewidth=2)
plt.tick_params(labelsize=18)
plt.Axes.set_xscale(plt.gca(), 'log')
plt.Axes.set_yscale(plt.gca(), 'log')
plt.xlabel('Radius [kpc]', size=20)
plt.ylabel(r"Density [$M_\odot$/kpc$^3$]", size=20)
plt.legend(fontsize=12)
```
